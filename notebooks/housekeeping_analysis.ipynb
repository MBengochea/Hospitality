```python
from pathlib import Path
import sys
ROOT=Path.cwd().parent if Path.cwd().name=="notebooks" else Path.cwd()
sys.path.insert(0,str(ROOT))
import pandas as pd
import matplotlib.pyplot as plt
from src.models import train_housekeeping_models
df=pd.read_csv(ROOT/"data/synthetic/housekeeping_operations.csv",parse_dates=["date"])
df.head()
```

```python
summary=df.groupby(["room_type","cleaning_type"]).agg(rooms=("room_number","count"),avg_minutes=("cleaning_time_minutes","mean"),reclean_rate=("reclean_required","mean"),pass_rate=("ready_by_13","mean"))
summary.round(3)
```

```python
summary.reset_index().pivot(index="room_type",columns="cleaning_type",values="avg_minutes").plot(kind="bar")
plt.ylabel("Average cleaning minutes")
plt.title("Synthetic cleaning time by room and cleaning type")
plt.tight_layout(); plt.show()
```
